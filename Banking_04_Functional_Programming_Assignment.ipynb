### Q33. Group accounts using a helper function for banking case 1; use sample reference `ACC-1033` and explain the result.

```python
# Write your solution here
def group_by_type(accounts):
    result = {}

    for account in accounts:
        account_type = account["type"]

        if account_type not in result:
            result[account_type] = []

        result[account_type].append(account)

    return result


accounts = [
    {"account": "ACC-1033A", "type": "Savings"},
    {"account": "ACC-1033B", "type": "Current"},
    {"account": "ACC-1033C", "type": "Savings"}
]

result = group_by_type(accounts)

print(result)
```

### Q35. Write a decorator that logs checkout calls for banking case 1; use sample reference `ACC-1035` and explain the result.

3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
from functools import wraps


def log_checkout(function):
    @wraps(function)
    def wrapper(*args, **kwargs):
        print("Checkout function called:", function.__name__)
        return function(*args, **kwargs)

    return wrapper


@log_checkout
def checkout(amount):
    return amount


result = checkout(1500)

print("Checkout Amount:", result)
```

### Q36. Build a map-filter-reduce transactions pipeline for banking case 1; use sample reference `ACC-1036` and explain the result.

**Hints**

1. First express the input and return value of one pure function.
2. Use `filter` for selection, `map` for transformation, and `reduce` only when a clearer built-in is not available.
3. Test the pipeline with an empty collection and avoid mutating account records in place.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
from functools import reduce

transactions = [
    {"amount": 1000, "approved": True},
    {"amount": 500, "approved": False},
    {"amount": 2000, "approved": True}
]

approved = filter(
    lambda transaction: transaction["approved"],
    transactions
)

amounts = map(
    lambda transaction: transaction["amount"],
    approved
)

total = reduce(
    lambda a, b: a + b,
    amounts,
    0
)

print("Approved Total:", total)
```

```python
# Write your solution here
from functools import wraps

def retry(times):

    def decorator(function):

        @wraps(function)
        def wrapper(*args, **kwargs):

            for attempt in range(1, times + 1):
                try:
                    return function(*args, **kwargs)

                except Exception as error:
                    print(f"Attempt {attempt} failed")

                    if attempt == times:
                        raise error

        return wrapper

    return decorator


attempts = 0


@retry(3)
def transfer():
    global attempts

    attempts += 1

    if attempts < 3:
        raise ConnectionError("Transfer failed")

    return "Transfer successful"


print(transfer())
```
