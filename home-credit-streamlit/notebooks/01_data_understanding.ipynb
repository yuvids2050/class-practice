# 01 — Data Understanding

Home Credit Default Risk: understand shape, structure, and relationships
across all 7 tables before any cleaning or feature engineering.

Tables: `application_train`, `bureau`, `bureau_balance`, `previous_application`,
`POS_CASH_balance`, `installments_payments`, `credit_card_balance`.

Relationship keys: `SK_ID_CURR` (customer), `SK_ID_PREV` (previous application),
`SK_ID_BUREAU` (bureau credit).

```python
import pandas as pd
import numpy as np

pd.set_option("display.max_columns", 50)

DATA_DIR = "../data"
app = pd.read_csv(f"{DATA_DIR}/application_train.csv")
bureau = pd.read_csv(f"{DATA_DIR}/bureau.csv")
bureau_balance = pd.read_csv(f"{DATA_DIR}/bureau_balance.csv")
previous_application = pd.read_csv(f"{DATA_DIR}/previous_application.csv")
pos_cash = pd.read_csv(f"{DATA_DIR}/POS_CASH_balance.csv")
installments = pd.read_csv(f"{DATA_DIR}/installments_payments.csv")
credit_card = pd.read_csv(f"{DATA_DIR}/credit_card_balance.csv")

tables = {
    "application_train": app, "bureau": bureau, "bureau_balance": bureau_balance,
    "previous_application": previous_application, "POS_CASH_balance": pos_cash,
    "installments_payments": installments, "credit_card_balance": credit_card,
}
for name, df in tables.items():
    print(f"{name:25s} shape={df.shape}")
```

## Table relationships

- `application_train.SK_ID_CURR` is the primary customer key.
- `bureau.SK_ID_CURR` links each external credit record back to a customer;
  `bureau.SK_ID_BUREAU` is that credit's own key.
- `bureau_balance.SK_ID_BUREAU` links monthly bureau status rows back to a
  bureau credit (and, through bureau, to a customer).
- `previous_application.SK_ID_CURR` links past Home Credit applications to a
  customer; `SK_ID_PREV` is that previous application's own key.
- `POS_CASH_balance`, `installments_payments`, and `credit_card_balance` all
  key on `SK_ID_PREV` (and carry `SK_ID_CURR` too) — monthly/event-level
  detail under each previous application.

```python
print("Customers in application_train:", app["SK_ID_CURR"].nunique())
print("Customers with >=1 bureau record:", bureau["SK_ID_CURR"].nunique())
print("Customers with >=1 previous application:", previous_application["SK_ID_CURR"].nunique())
print("Bureau credits with balance history:", bureau_balance["SK_ID_BUREAU"].nunique(), "/", bureau["SK_ID_BUREAU"].nunique())
```
