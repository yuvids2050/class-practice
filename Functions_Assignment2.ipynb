```python
#1 Using filter() and map():

# Extract all even numbers.
# Calculate the square of every even number.
# Return the squared values as a list.
numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

result = list(
    map(
        lambda x: x ** 2,
        filter(lambda x: x % 2 == 0, numbers)
    )
)

print(result)
```

```python
#2 Using filter() and reduce():

# Extract all odd numbers.
# Calculate their total sum.
from functools import reduce

numbers = [12, 45, 67, 89, 23, 56, 91, 34, 78, 100]

result = reduce(
    lambda x, y: x + y,
    filter(lambda x: x % 2 != 0, numbers)
)

print(result)
```

```python
#3 Using filter() and map():

# Extract numbers greater than 50.
# Multiply every selected number by 2.
numbers = [10, 25, 40, 55, 70, 85, 100]

result = list(
    map(
        lambda x: x * 2,
        filter(lambda x: x > 50, numbers)
    )
)

print(result)
```

```python
#4 Using filter(), map(), and reduce():

# Extract positive numbers.
# Calculate the square of every positive number.
# Calculate the total sum of the squared values.
from functools import reduce

numbers = [-15, 20, -35, 40, 50, -10, 60]

result = reduce(
    lambda x, y: x + y,
    map(
        lambda x: x ** 2,
        filter(lambda x: x > 0, numbers)
    )
)

print(result)
```

```python
#5 Using filter() and map():

# Extract words containing more than five characters.
# Convert the selected words to uppercase.
words = ["python", "azure", "ai", "developer", "cloud", "analytics", "spark"]

result = list(
    map(
        lambda x: x.upper(),
        filter(lambda x: len(x) > 5, words)
    )
)

print(result)
```

```python
#6 Using filter() and map():

# Extract names beginning with "A".
# Convert the selected names to lowercase.
names = ["Anil", "Rahul", "Ajay", "Meena", "Arjun", "Kiran", "Akhil"]

result = list(
    map(
        lambda x: x.lower(),
        filter(lambda x: x.startswith("A"), names)
    )
)

print(result)
```

```python
#7 Using filter() and reduce():

# Extract numbers divisible by both 3 and 5.
# Calculate their total sum.
from functools import reduce

numbers = [3, 5, 10, 15, 18, 30, 42, 45, 60, 75]

result = reduce(
    lambda x, y: x + y,
    filter(lambda x: x % 3 == 0 and x % 5 == 0, numbers)
)

print(result)
```

```python
#8 Using filter() and reduce():

# Extract palindrome words.
# Join them into one string separated by " | ".
from functools import reduce

words = ["madam", "python", "level", "azure", "radar", "cloud", "civic"]

result = reduce(
    lambda x, y: x + " | " + y,
    filter(lambda x: x == x[::-1], words)
)

print(result)
```

```python
#9 Using filter(), map(), and reduce():

# Extract temperatures greater than 30°C.
# Convert the selected values into Fahrenheit.
# Calculate the average Fahrenheit temperature
from functools import reduce

temperatures = [18, 24, 30, 35, 40, 12, 28, 33, 15, 42]

fahrenheit = list(
    map(
        lambda c: c * 9 / 5 + 32,
        filter(lambda c: c > 30, temperatures)
    )
)

total = reduce(lambda x, y: x + y, fahrenheit)

average = total / len(fahrenheit)

print(average)
```

```python
#12 Using filter() and reduce():

# Extract employees from the IT department.
# Calculate the total salary of IT employees.
from functools import reduce

employees = [
    {"id": 101, "name": "Rahul", "department": "IT", "salary": 45000},
    {"id": 102, "name": "Anil", "department": "HR", "salary": 65000},
    {"id": 103, "name": "Sneha", "department": "IT", "salary": 72000},
    {"id": 104, "name": "Kiran", "department": "IT", "salary": 82000},
    {"id": 105, "name": "Meena", "department": "Sales", "salary": 55000}
]

result = reduce(
    lambda total, emp: total + emp["salary"],
    filter(lambda emp: emp["department"] == "IT", employees),
    0
)

print(result)
```

```python
#13 Using filter(), map(), and reduce():

# Extract active employees.
# Increase every active employee's salary by 15%.
# Calculate the total updated payroll.
from functools import reduce

employees = [
    {"id": 101, "name": "Rahul", "salary": 45000, "active": True},
    {"id": 102, "name": "Anil", "salary": 65000, "active": True},
    {"id": 103, "name": "Sneha", "salary": 72000, "active": False},
    {"id": 104, "name": "Kiran", "salary": 82000, "active": True},
    {"id": 105, "name": "Meena", "salary": 55000, "active": True},
    {"id": 106, "name": "Arjun", "salary": 98000, "active": False}
]

result = reduce(
    lambda x, y: x + y,
    map(
        lambda emp: emp["salary"] * 1.15,
        filter(lambda emp: emp["active"], employees)
    )
)

print(result)
```

```python
#14 Using filter() and reduce():

# Extract students who passed.
# Find the topper among the passed students.
from functools import reduce

students = [
    {"name": "Ravi", "marks": 92, "passed": True},
    {"name": "Kiran", "marks": 68, "passed": True},
    {"name": "Meena", "marks": 85, "passed": True},
    {"name": "Ajay", "marks": 48, "passed": True},
    {"name": "Sneha", "marks": 35, "passed": False},
    {"name": "Rahul", "marks": 95, "passed": True}
]

result = reduce(
    lambda x, y: x if x["marks"] > y["marks"] else y,
    filter(lambda student: student["passed"], students)
)

print(result)
```

```python
#15 Using filter(), map(), and reduce():

# Extract students scoring more than 70.
# Extract only their marks.
# Calculate the average marks.
from functools import reduce

students = [
    {"name": "Ravi", "marks": 92},
    {"name": "Kiran", "marks": 68},
    {"name": "Meena", "marks": 85},
    {"name": "Ajay", "marks": 48},
    {"name": "Sneha", "marks": 79},
    {"name": "Rahul", "marks": 95}
]

marks = list(
    map(
        lambda student: student["marks"],
        filter(lambda student: student["marks"] > 70, students)
    )
)

average = reduce(lambda x, y: x + y, marks) / len(marks)

print(average)
```

```python
#16 Using filter(), map(), and reduce():

# Extract products currently in stock.
# Calculate each product's inventory value using price × stock.
# Calculate the total inventory value.
from functools import reduce

products = [
    {"name": "Laptop", "price": 55000, "stock": 5},
    {"name": "Mouse", "price": 600, "stock": 25},
    {"name": "Keyboard", "price": 1500, "stock": 0},
    {"name": "Monitor", "price": 18000, "stock": 8},
    {"name": "Printer", "price": 12000, "stock": 0},
    {"name": "SSD", "price": 6500, "stock": 12}
]

result = reduce(
    lambda x, y: x + y,
    map(
        lambda p: p["price"] * p["stock"],
        filter(lambda p: p["stock"] > 0, products)
    )
)

print(result)
```

```python
#19 Using filter(), map(), and reduce():

# Extract transactions whose amount is greater than 5000.
# Add 18% GST to every selected transaction.
# Calculate the grand total after GST.
from functools import reduce

transactions = [
    {"id": 1, "amount": 1200},
    {"id": 2, "amount": 5200},
    {"id": 3, "amount": 18000},
    {"id": 4, "amount": 2500},
    {"id": 5, "amount": 9600},
    {"id": 6, "amount": 7200}
]

result = reduce(
    lambda x, y: x + y,
    map(
        lambda t: t["amount"] * 1.18,
        filter(lambda t: t["amount"] > 5000, transactions)
    )
)

print(result)
```

```python
#20 Using filter(), map(), and reduce():

# Extract active salespeople from the South region.
# Calculate a 5% incentive for each selected salesperson.
# Create a list containing salesperson name, sales amount, and incentive.
# Calculate the total incentive payable.
from functools import reduce

sales_records = [
    {"salesperson": "Rahul", "region": "South", "sales": 45000, "active": True},
    {"salesperson": "Anil", "region": "North", "sales": 65000, "active": True},
    {"salesperson": "Sneha", "region": "South", "sales": 72000, "active": False},
    {"salesperson": "Kiran", "region": "South", "sales": 82000, "active": True},
    {"salesperson": "Meena", "region": "West", "sales": 55000, "active": True},
    {"salesperson": "Arjun", "region": "South", "sales": 98000, "active": True}
]

selected = list(
    map(
        lambda s: {
            "salesperson": s["salesperson"],
            "sales": s["sales"],
            "incentive": s["sales"] * 0.05
        },
        filter(
            lambda s: s["region"] == "South" and s["active"],
            sales_records
        )
    )
)

print(selected)

total_incentive = reduce(
    lambda total, s: total + s["incentive"],
    selected,
    0
)

print(total_incentive)
```
