```python
# Write your solution here
class Account:
    def __init__(self, account_ref, balance):
        self.account_ref = account_ref
        self.balance = balance

    def __str__(self):
        return f"{self.account_ref}: ₹{self.balance}"


class UnitOfWork:
    def __init__(self, accounts):
        self.accounts = accounts
        self.original_balances = {}

    def __enter__(self):
        for ref, account in self.accounts.items():
            self.original_balances[ref] = account.balance

        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type is not None:
            for ref, balance in self.original_balances.items():
                self.accounts[ref].balance = balance

            print("Transaction rolled back")
            return False

        print("Transaction committed")
        return False


accounts = {
    "ACC-1075": Account("ACC-1075", 10000),
    "ACC-1076": Account("ACC-1076", 5000)
}

try:
    with UnitOfWork(accounts):
        accounts["ACC-1075"].balance -= 2000
        accounts["ACC-1076"].balance += 2000

except Exception:
    print("Transaction failed")

for account in accounts.values():
    print(account)
```

### Q85. Implement a unit-of-work transaction for banking case 2; use sample reference `ACC-1085` and explain the result.

2. Choose composition for has-a relationships; use inheritance only for genuine is-a substitution.

```python
# Write your solution here
class Account:
    def __init__(self, account_ref, balance):
        self.account_ref = account_ref
        self.balance = balance


class UnitOfWork:
    def __init__(self, accounts):
        self.accounts = accounts
        self.original_balances = {}

    def __enter__(self):
        for ref, account in self.accounts.items():
            self.original_balances[ref] = account.balance

        return self

    def __exit__(self, exc_type, exc_value, traceback):
        if exc_type is not None:
            for ref, balance in self.original_balances.items():
                self.accounts[ref].balance = balance

            print("Transaction rolled back")
        else:
            print("Transaction committed")

        return False


accounts = {
    "ACC-1085": Account("ACC-1085", 10000),
    "ACC-1086": Account("ACC-1086", 5000)
}

with UnitOfWork(accounts):
    accounts["ACC-1085"].balance -= 2000
    accounts["ACC-1086"].balance += 2000

print(accounts["ACC-1085"].balance)
print(accounts["ACC-1086"].balance)
```
