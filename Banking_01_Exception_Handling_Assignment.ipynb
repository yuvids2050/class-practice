### Q33. Process transfers while isolating bad rows for banking case 1; use sample reference `ACC-1033` and explain the result.

**Hints**

1. Identify the smallest risky statement and the narrow exception it can raise.
2. Keep business-rule failures distinct from programming or infrastructure failures.
3. For medium/difficult work, preserve the original cause and protect partially updated balances.

**Edge cases to test:** empty input, invalid types, boundary values, and a realistic failure path.

```python
# Write your solution here
transfers = [
    ("A101", "A102", 500),
    ("A103", "A104", "bad"),
    ("A105", "A106", 1000)
]

successful = []
failed = []

for source, destination, amount in transfers:

    try:
        amount = float(amount)

        if amount <= 0:
            raise ValueError("Transfer amount must be positive.")

    except (ValueError, TypeError) as e:
        failed.append((source, destination, str(e)))

    else:
        successful.append((source, destination, amount))


print("Account reference:", "ACC-1033")
print("Successful transfers:", successful)
print("Failed transfers:", failed)
```

### Q100. Build an exception-safe reconciliation pipeline for banking case 3; use sample reference `ACC-1100` and explain the result.

1. Identify the smallest risky statement and the narrow exception it can raise.

3. For medium/difficult work, preserve the original cause and protect partially updated balances.

```python
# Write your solution here
transactions = [
    {"id": 1, "bank": 2000, "ledger": 2000},
    {"id": 2, "bank": 3500, "ledger": 3000},
    {"id": 3, "bank": "invalid", "ledger": 800}
]

reconciled = []
failed = []

for transaction in transactions:

    try:
        bank_amount = float(transaction["bank"])
        ledger_amount = float(transaction["ledger"])

        difference = bank_amount - ledger_amount

    except (ValueError, TypeError, KeyError) as e:

        failed.append({
            "id": transaction.get("id"),
            "success": False,
            "error": str(e)
        })

    else:

        reconciled.append({
            "id": transaction["id"],
            "success": True,
            "difference": difference
        })


print("Account:", "ACC-1100")

print("Reconciled:")
for item in reconciled:
    print(item)

print("\nFailed:")
for item in failed:
    print(item)
```
