```python
import numpy as np

# Store names
stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

# Product categories
categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

# Sales revenue in rupees
# Rows represent stores
# Columns represent product categories

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

# Cost incurred in rupees
# Rows represent stores
# Columns represent product categories

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

print("Stores:")
print(stores)

print("\nProduct Categories:")
print(categories)

print("\nSales Data:")
print(sales_data)

print("\nCost Data:")
print(cost_data)
```

```python
# Question 40: Low-Margin Transactions
# Management considers a profit margin below 22% to be weak.

# Find all store-category combinations where the profit margin is below 22%.

# Display:

# Store
# Category
# Sales
# Cost
# Profit
# Profit margin
# Use Boolean indexing.
profit_data = sales_data - cost_data

profit_margin = (profit_data / sales_data) * 100

condition = profit_margin < 22

rows, columns = np.where(condition)

for row, column in zip(rows, columns):
    print(
        "Store:", stores[row],
        "| Category:", categories[column],
        "| Sales:", sales_data[row, column],
        "| Cost:", cost_data[row, column],
        "| Profit:", profit_data[row, column],
        "| Margin:", round(profit_margin[row, column], 2), "%"
    )
```

```python
discount_rates = np.array([
    0.10,  # Electronics: 10%
    0.05,  # Groceries: 5%
    0.15,  # Clothing: 15%
    0.12,  # Furniture: 12%
    0.08,  # Beauty: 8%
    0.10   # Sports: 10%
])
```

```python
# Question 58: Discount Amount Matrix
# Calculate the discount amount for each store-category combination.

# Use broadcasting.
discount_amount = sales_data * discount_rates

print(discount_amount)
```

```python
# Question 59: Sales After Discount
# Calculate the net sales value after applying the category-wise discounts
sales_after_discount = sales_data - discount_amount

print(sales_after_discount)
```

```python
# Question 60: Profit After Discount
# Assuming costs remain unchanged, calculate the new profit after the discounts are applied.
profit_after_discount = sales_after_discount - cost_data

print(profit_after_discount)
```

```python
# Question 61: Profit Margin After Discount
# Calculate the new profit margin after applying discounts.
profit_margin_after_discount = (
    profit_after_discount / sales_after_discount
) * 100

profit_margin_after_discount = np.round(
    profit_margin_after_discount, 2
)

print(profit_margin_after_discount)
```

```python
# Question 64: Store Most Affected by Discounts
# Identify the store that loses the highest amount of revenue due to discounts.
store_discount_loss = np.sum(discount_amount, axis=1)

best_affected_index = np.argmax(store_discount_loss)

print("Store:", stores[best_affected_index])
print(
    "Revenue lost:",
    f"₹{store_discount_loss[best_affected_index]:,.2f}"
)
```

```python
# Part 20: Final Retail Analytics Project
# Prepare a complete NumPy-based retail performance report.

# Your report must contain the following information:

# Company Summary
# Total company sales
# Total company cost
# Total company profit
# Overall company profit margin
# Average transaction value
# Highest individual sales value
# Lowest individual sales value
# Store Summary
# For every store, display:

# Total sales
# Total cost
# Total profit
# Profit margin
# Number of category targets achieved
# Total units sold
# Expansion eligibility
# Category Summary
# For every category, display:

# Total sales
# Total cost
# Total profit
# Profit margin
# Average sales
# Number of stores achieving the target
# Total units sold
# Management Recommendations
# Based on the analysis, identify:

# Best-performing store
# Lowest-performing store
# Most profitable category
# Lowest-profit category
# Store recommended for expansion
# Category requiring additional marketing
# Store-category combinations needing cost optimization
# Store-category combinations requiring inventory reorder
# Expected Final Output Format
# SMARTMART RETAIL PERFORMANCE REPORT
# ===================================

# COMPANY SUMMARY
# ---------------
# Total Sales:
# Total Cost:
# Total Profit:
# Overall Profit Margin:

# BEST-PERFORMING STORE
# ---------------------
# Store:
# Sales:
# Cost:
# Profit:
# Profit Margin:

# MOST PROFITABLE CATEGORY
# ------------------------
# Category:
# Sales:
# Cost:
# Profit:
# Profit Margin:

# EXPANSION RECOMMENDATIONS
# -------------------------
# Eligible Stores:

# COST OPTIMIZATION
# -----------------
# Store:
# Category:
# Cost:
# Profit Margin:

# INVENTORY REORDER
# -----------------
# Store:
# Category:
# Closing Stock:
# Reorder Quantity:

import numpy as np

# ============================================================
# DATA
# ============================================================

stores = np.array([
    "Bengaluru",
    "Hyderabad",
    "Chennai",
    "Mumbai",
    "Pune"
])

categories = np.array([
    "Electronics",
    "Groceries",
    "Clothing",
    "Furniture",
    "Beauty",
    "Sports"
])

sales_data = np.array([
    [850000, 420000, 310000, 275000, 190000, 225000],
    [780000, 460000, 295000, 310000, 210000, 240000],
    [920000, 390000, 340000, 260000, 185000, 215000],
    [1050000, 510000, 375000, 345000, 250000, 290000],
    [690000, 405000, 280000, 225000, 175000, 205000]
])

cost_data = np.array([
    [640000, 315000, 220000, 205000, 125000, 160000],
    [590000, 350000, 210000, 225000, 140000, 170000],
    [700000, 295000, 245000, 195000, 120000, 155000],
    [790000, 390000, 265000, 250000, 165000, 205000],
    [530000, 310000, 205000, 170000, 115000, 145000]
])

sales_targets = np.array([
    800000,  # Electronics
    450000,  # Groceries
    320000,  # Clothing
    300000,  # Furniture
    200000,  # Beauty
    230000   # Sports
])

opening_stock = np.array([
    [120, 450, 300, 90, 250, 180],
    [110, 500, 280, 100, 270, 200],
    [130, 420, 320, 85, 240, 170],
    [150, 550, 350, 120, 300, 220],
    [100, 400, 260, 75, 220, 160]
])

units_sold = np.array([
    [95, 380, 240, 65, 190, 140],
    [85, 420, 230, 75, 210, 155],
    [105, 350, 270, 60, 180, 130],
    [125, 470, 295, 90, 235, 175],
    [75, 330, 205, 55, 165, 120]
])

# ============================================================
# CORE CALCULATIONS
# ============================================================

profit_data = sales_data - cost_data
profit_margin = (profit_data / sales_data) * 100

store_sales = np.sum(sales_data, axis=1)
store_cost = np.sum(cost_data, axis=1)
store_profit = np.sum(profit_data, axis=1)
store_margin = (store_profit / store_sales) * 100

category_sales = np.sum(sales_data, axis=0)
category_cost = np.sum(cost_data, axis=0)
category_profit = np.sum(profit_data, axis=0)
category_margin = (category_profit / category_sales) * 100
category_average_sales = np.mean(sales_data, axis=0)

target_achieved = sales_data >= sales_targets
store_target_count = np.sum(target_achieved, axis=1)
category_target_count = np.sum(target_achieved, axis=0)

closing_stock = opening_stock - units_sold
reorder_quantity = np.where(closing_stock < 25, 100 - closing_stock, 0)
reorder_needed = closing_stock < 25

# ============================================================
# HELPERS
# ============================================================

def money(value):
    return f"₹{value:,.2f}"

def show_pair(title, value):
    print(f"{title}: {value}")

# ============================================================
# COMPANY SUMMARY
# ============================================================

total_company_sales = np.sum(sales_data)
total_company_cost = np.sum(cost_data)
total_company_profit = np.sum(profit_data)
overall_company_profit_margin = (total_company_profit / total_company_sales) * 100
average_transaction_value = np.mean(sales_data)
highest_individual_sales_value = np.max(sales_data)
lowest_individual_sales_value = np.min(sales_data)

# ============================================================
# MANAGEMENT RECOMMENDATIONS
# ============================================================

best_store_index = np.argmax(store_profit)
lowest_store_index = np.argmin(store_profit)

most_profitable_category_index = np.argmax(category_profit)
lowest_profit_category_index = np.argmin(category_profit)

expansion_eligible = (
    (store_sales > 2500000) &
    (store_profit > 700000) &
    (store_margin > 25)
)

promotion_category = (
    (category_average_sales < 300000) &
    (category_margin > 25)
)

cost_optimization_mask = (
    (cost_data > 250000) &
    (profit_margin < 25)
)

reorder_mask = closing_stock < 25

# ============================================================
# FINAL REPORT
# ============================================================

print("SMARTMART RETAIL PERFORMANCE REPORT")
print("===================================")

print("\nCOMPANY SUMMARY")
print("---------------")
show_pair("Total Sales", money(total_company_sales))
show_pair("Total Cost", money(total_company_cost))
show_pair("Total Profit", money(total_company_profit))
show_pair("Overall Profit Margin", f"{overall_company_profit_margin:.2f}%")
show_pair("Average Transaction Value", money(average_transaction_value))
show_pair("Highest Individual Sales Value", money(highest_individual_sales_value))
show_pair("Lowest Individual Sales Value", money(lowest_individual_sales_value))

print("\nSTORE SUMMARY")
print("-------------")
for i in range(len(stores)):
    show_pair("Store", stores[i])
    show_pair("Total Sales", money(store_sales[i]))
    show_pair("Total Cost", money(store_cost[i]))
    show_pair("Total Profit", money(store_profit[i]))
    show_pair("Profit Margin", f"{store_margin[i]:.2f}%")
    show_pair("Number of Category Targets Achieved", int(store_target_count[i]))
    show_pair("Total Units Sold", int(np.sum(units_sold[i])))
    show_pair("Expansion Eligibility", "Yes" if expansion_eligible[i] else "No")
    print()

print("\nCATEGORY SUMMARY")
print("----------------")
for i in range(len(categories)):
    show_pair("Category", categories[i])
    show_pair("Total Sales", money(category_sales[i]))
    show_pair("Total Cost", money(category_cost[i]))
    show_pair("Total Profit", money(category_profit[i]))
    show_pair("Profit Margin", f"{category_margin[i]:.2f}%")
    show_pair("Average Sales", money(category_average_sales[i]))
    show_pair("Number of Stores Achieving Target", int(category_target_count[i]))
    show_pair("Total Units Sold", int(np.sum(units_sold[:, i])))
    print()

print("\nMANAGEMENT RECOMMENDATIONS")
print("--------------------------")
show_pair("Best-Performing Store", stores[best_store_index])
show_pair("Lowest-Performing Store", stores[lowest_store_index])
show_pair("Most Profitable Category", categories[most_profitable_category_index])
show_pair("Lowest-Profit Category", categories[lowest_profit_category_index])

print("\nStore Recommended for Expansion:")
for i in np.where(expansion_eligible)[0]:
    print(f"- {stores[i]}")

print("\nCategory Requiring Additional Marketing:")
for i in np.where(promotion_category)[0]:
    print(f"- {categories[i]}")

print("\nStore-Category Combinations Needing Cost Optimization:")
for row, col in np.argwhere(cost_optimization_mask):
    print(
        f"- {stores[row]} | {categories[col]} | "
        f"Cost: {money(cost_data[row, col])} | "
        f"Profit Margin: {profit_margin[row, col]:.2f}%"
    )

print("\nStore-Category Combinations Requiring Inventory Reorder:")
for row, col in np.argwhere(reorder_mask):
    print(
        f"- {stores[row]} | {categories[col]} | "
        f"Closing Stock: {int(closing_stock[row, col])} | "
        f"Reorder Quantity: {int(reorder_quantity[row, col])}"
    )

```
