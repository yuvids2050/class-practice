```python
# Assignment 13: Factorial Using a Function
# Write a function named calculate_factorial().

# Requirements
# Accept a non-negative integer.
# Calculate its factorial using a loop.
# Do not use math.factorial().
# Return the factorial.
# Handle negative input.
# Example Input
# Enter a number: 5
# Expected Output
# Factorial: 120
def calculate_factorial(number):

    if number < 0:
        return "Factorial does not exist for negative numbers."

    factorial = 1

    for i in range(1, number + 1):
        factorial *= i

    return factorial


number = int(input("Enter a number: "))

result = calculate_factorial(number)

print("Factorial:", result)
```

```python
# Assignment 32: Function Returning a Dictionary
# Write a function named analyze_text().

# Requirements
# Accept a string.
# Return a dictionary containing:
# Total characters
# Total words
# Total vowels
# Total digits
# Total spaces
# Display every result clearly.
def analyze_text(text):

    vowels = "aeiouAEIOU"

    result = {
        "Total Characters": len(text),
        "Total Words": len(text.split()),
        "Total Vowels": 0,
        "Total Digits": 0,
        "Total Spaces": 0
    }

    for ch in text:

        if ch in vowels:
            result["Total Vowels"] += 1

        if ch.isdigit():
            result["Total Digits"] += 1

        if ch == " ":
            result["Total Spaces"] += 1

    return result


text = input("Enter text: ")

analysis = analyze_text(text)

for key, value in analysis.items():
    print(key, ":", value)
```

```python
# Assignment 34: Prime Number Function
# Write a function named is_prime().

# Requirements
# Accept one integer.
# Return True when the number is prime.
# Return False otherwise.
# Use the function to print all prime numbers from 1 to 100.
# Hint
# A prime number has exactly two factors.
def is_prime(number):

    if number <= 1:
        return False

    for i in range(2, number):

        if number % i == 0:
            return False

    return True


print("Prime Numbers from 1 to 100")

for number in range(1, 101):

    if is_prime(number):
        print(number, end=" ")
```

```python
# Assignment 35: Fibonacci Function
# Write a function named generate_fibonacci().

# Requirements
# Accept the number of terms.
# Return the complete Fibonacci series as a list.
# Handle zero and negative term counts.
# Do not use recursion in this assignment.
# Example Input
# Enter number of terms: 7
# Expected Output
# [0, 1, 1, 2, 3, 5, 8]
def generate_fibonacci(terms):

    if terms < 0:
        return "Number of terms cannot be negative."

    if terms == 0:
        return []

    if terms == 1:
        return [0]

    fibonacci = [0, 1]

    while len(fibonacci) < terms:

        next_number = fibonacci[-1] + fibonacci[-2]

        fibonacci.append(next_number)

    return fibonacci


terms = int(input("Enter number of terms: "))

result = generate_fibonacci(terms)

print(result)
```

```python
# Assignment 56: Mask Email Addresses
# Given a list of email addresses, mask the username portion.

# Example
# bharath@gmail.com → b******@gmail.com
# Requirements
# Use map().
# Keep the first character of the username.
# Replace remaining username characters with *.
# Keep the domain unchanged.
# Create a separate named function for masking.

def mask_email(email):

    if "@" not in email:
        return "Invalid Email"

    username, domain = email.split("@")

    if len(username) <= 1:
        return email

    masked = username[0] + "*" * (len(username) - 1)

    return masked + "@" + domain


emails = input("Enter email addresses separated by spaces: ").split()

masked_emails = list(map(mask_email, emails))

print("Original Emails")
print(emails)

print()

print("Masked Emails")
print(masked_emails)
```

```python
# Assignment 81: Calculate Factorial Using reduce()
# Write a function that calculates factorial using reduce().

# Requirements
# Accept a non-negative integer.
# Generate numbers from 1 to the given number.
# Multiply them using reduce().
# Return 1 when the input is 0.
# Reject negative values.
from functools import reduce

def calculate_factorial(number):

    if number < 0:
        return "Factorial is not defined for negative numbers."

    if number == 0:
        return 1

    numbers = range(1, number + 1)

    return reduce(
        lambda accumulator, value: accumulator * value,
        numbers,
        1
    )


number = int(input("Enter a number: "))

print("Factorial:", calculate_factorial(number))
```

```python
# Assignment 92: Closure-Based Bank Balance
# Write a function named create_bank_account().

# Requirements
# Store the balance in the outer function.
# Define inner functions for:
# Deposit
# Withdrawal
# Balance enquiry
# Use nonlocal to update the balance.
# Reject withdrawals greater than the available balance.
# Return all three inner functions.
# Hint
# Return multiple functions:

# return deposit, withdraw, check_balance
def create_bank_account(balance):

    def deposit(amount):

        nonlocal balance

        balance += amount

        print("Deposited:", amount)

    def withdraw(amount):

        nonlocal balance

        if amount > balance:

            print("Insufficient Balance")

        else:

            balance -= amount

            print("Withdrawn:", amount)

    def check_balance():

        print("Current Balance:", balance)

    return deposit, withdraw, check_balance


deposit, withdraw, check_balance = create_bank_account(10000)

check_balance()

deposit(2500)

check_balance()

withdraw(3000)

check_balance()

withdraw(15000)

check_balance()
```

```python
# Assignment 100: Complete Functional Programming Project
# Develop a complete Employee Performance Analysis System using functions, lambda expressions, map(), filter(), reduce(), nested functions, closures, and LEGB scope.

# Sample Employee Data
# employees = [
#     {
#         "id": 101,
#         "name": "Ravi",
#         "department": "IT",
#         "salary": 60000,
#         "performance_score": 88,
#         "experience": 5
#     },
#     {
#         "id": 102,
#         "name": "Anitha",
#         "department": "HR",
#         "salary": 45000,
#         "performance_score": 72,
#         "experience": 4
#     },
#     {
#         "id": 103,
#         "name": "Kiran",
#         "department": "IT",
#         "salary": 75000,
#         "performance_score": 94,
#         "experience": 7
#     },
#     {
#         "id": 104,
#         "name": "Meena",
#         "department": "Finance",
#         "salary": 55000,
#         "performance_score": 81,
#         "experience": 6
#     },
#     {
#         "id": 105,
#         "name": "Arjun",
#         "department": "IT",
#         "salary": 40000,
#         "performance_score": 65,
#         "experience": 2
#     }
# ]
# Functional Requirements
# Part 1: Display Employees
# Write a function named display_employees() that prints all employee details in a readable format.

# Part 2: Extract Employee Names
# Use map() to extract employee names.

# Part 3: Filter High Performers
# Use filter() to select employees whose performance score is 80 or above.

# Part 4: Filter by Department
# Write a function that accepts a department name and returns matching employees using filter().

# Part 5: Salary Increase
# Use map() to calculate revised salaries based on these conditions:

# Performance score 90 or above: 15% increase
# Performance score 80–89: 10% increase
# Performance score 70–79: 5% increase
# Below 70: No increase
# Do not modify the original employee dictionaries directly.

# Part 6: Total Salary Expense
# Use reduce() to calculate:

# Current total salary
# Revised total salary
# Total additional salary expense
# Part 7: Highest-Paid Employee
# Use reduce() to find the employee with the highest salary.

# Do not use max().

# Part 8: Sort Employees
# Use sorted() with lambda expressions to sort employees by:

# Name
# Salary
# Performance score
# Experience
# Part 9: Closure-Based Bonus Calculator
# Create a function named create_bonus_calculator().

# The outer function should accept a bonus percentage.
# The inner function should accept a salary.
# Return the calculated bonus.
# Create separate bonus calculators for 5%, 10%, and 15%.
# Part 10: Nested Report Generator
# Create an outer function named generate_department_report().

# Inside it, define nested functions to calculate:

# Number of employees
# Total salary
# Average salary
# Average performance score
# Return the complete department report as a dictionary.

# Part 11: LEGB Demonstration
# Create a global variable:

# company_name = "Edukron"
# Inside the report function:

# Access the global company name.
# Create an enclosing variable for the department.
# Create a local variable for the report title.
# Add comments explaining how Python resolves each variable using LEGB.
# Part 12: Menu-Driven Program
# Create a menu containing the following options:

# 1. Display all employees
# 2. Display employee names
# 3. Display high performers
# 4. Search by department
# 5. Calculate revised salaries
# 6. Calculate total salary expense
# 7. Find highest-paid employee
# 8. Sort employees
# 9. Generate department report
# 10. Exit
from functools import reduce

# Global variable for LEGB demonstration
company_name = "Edukron"

employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]


def print_separator():
    print("-" * 110)


def display_employees(employee_list):
    """Print employee details in a readable format."""
    if not employee_list:
        print("No employees found.")
        return

    print_separator()
    for emp in employee_list:
        line = (
            f"ID: {emp['id']} | "
            f"Name: {emp['name']} | "
            f"Department: {emp['department']} | "
            f"Salary: {emp['salary']} | "
            f"Performance: {emp['performance_score']} | "
            f"Experience: {emp['experience']} years"
        )

        if "revised_salary" in emp:
            line += f" | Revised Salary: {emp['revised_salary']}"

        print(line)
    print_separator()


def get_employee_names(employee_list):
    """Use map() to extract employee names."""
    return list(map(lambda emp: emp["name"], employee_list))


def get_high_performers(employee_list):
    """Use filter() to select employees with performance score 80 or above."""
    return list(filter(lambda emp: emp["performance_score"] >= 80, employee_list))


def filter_by_department(employee_list, department_name):
    """Return employees matching the given department."""
    return list(
        filter(
            lambda emp: emp["department"].lower() == department_name.lower(),
            employee_list
        )
    )


def calculate_revised_salary(emp):
    """
    Create a new dictionary with revised salary based on performance score.
    Do not modify the original employee dictionary.
    """
    score = emp["performance_score"]

    if score >= 90:
        increase_rate = 0.15
    elif score >= 80:
        increase_rate = 0.10
    elif score >= 70:
        increase_rate = 0.05
    else:
        increase_rate = 0.00

    revised_salary = round(emp["salary"] * (1 + increase_rate), 2)
    increment_amount = round(revised_salary - emp["salary"], 2)

    return {
        **emp,
        "revised_salary": revised_salary,
        "increment_amount": increment_amount
    }


def get_revised_salaries(employee_list):
    """Use map() to calculate revised salaries for all employees."""
    return list(map(calculate_revised_salary, employee_list))


def get_current_total_salary(employee_list):
    """Use reduce() to calculate current total salary."""
    return reduce(lambda acc, emp: acc + emp["salary"], employee_list, 0)


def get_revised_total_salary(revised_employee_list):
    """Use reduce() to calculate revised total salary."""
    return reduce(lambda acc, emp: acc + emp["revised_salary"], revised_employee_list, 0)


def get_total_additional_expense(employee_list):
    """Current total vs revised total."""
    revised_list = get_revised_salaries(employee_list)
    current_total = get_current_total_salary(employee_list)
    revised_total = get_revised_total_salary(revised_list)
    additional = round(revised_total - current_total, 2)
    return current_total, revised_total, additional


def get_highest_paid_employee(employee_list):
    """Use reduce() to find the employee with the highest salary."""
    if not employee_list:
        return None

    return reduce(
        lambda a, b: a if a["salary"] >= b["salary"] else b,
        employee_list
    )


def sort_employees(employee_list):
    """Sort employees by name, salary, performance score, and experience."""
    by_name = sorted(employee_list, key=lambda emp: emp["name"])
    by_salary = sorted(employee_list, key=lambda emp: emp["salary"])
    by_performance = sorted(employee_list, key=lambda emp: emp["performance_score"])
    by_experience = sorted(employee_list, key=lambda emp: emp["experience"])
    return by_name, by_salary, by_performance, by_experience


def create_bonus_calculator(bonus_percentage):
    """
    Closure-based bonus calculator.
    The inner function remembers bonus_percentage from the enclosing scope.
    """
    def calculate_bonus(salary):
        return round(salary * bonus_percentage / 100, 2)

    return calculate_bonus


def show_bonus_demo(employee_list):
    """Demonstrate closure-based bonus calculators."""
    bonus_5 = create_bonus_calculator(5)
    bonus_10 = create_bonus_calculator(10)
    bonus_15 = create_bonus_calculator(15)

    print("\nClosure-Based Bonus Calculator Demo")
    print_separator()
    for emp in employee_list:
        print(
            f"{emp['name']} | Salary: {emp['salary']} | "
            f"5%: {bonus_5(emp['salary'])} | "
            f"10%: {bonus_10(emp['salary'])} | "
            f"15%: {bonus_15(emp['salary'])}"
        )
    print_separator()


def generate_department_report(employee_list, department):
    """
    Nested report generator.
    LEGB notes:
    - company_name is read from GLOBAL scope.
    - department is an ENCLSOING variable.
    - report_title is a LOCAL variable inside this function.
    """
    # Global scope access
    global company_name

    filtered_employees = list(
        filter(lambda emp: emp["department"].lower() == department.lower(), employee_list)
    )

    # Enclosing variable
    report_department = department.strip().title()

    # Local variable
    report_title = f"{company_name} | {report_department} Department Report"

    def count_employees():
        return len(filtered_employees)

    def total_salary():
        return reduce(lambda acc, emp: acc + emp["salary"], filtered_employees, 0)

    def average_salary():
        count = count_employees()
        if count == 0:
            return 0
        return round(total_salary() / count, 2)

    def average_performance_score():
        count = count_employees()
        if count == 0:
            return 0
        total_score = reduce(lambda acc, emp: acc + emp["performance_score"], filtered_employees, 0)
        return round(total_score / count, 2)

    report = {
        "company_name": company_name,
        "report_title": report_title,
        "department": report_department,
        "employee_count": count_employees(),
        "total_salary": total_salary(),
        "average_salary": average_salary(),
        "average_performance_score": average_performance_score(),
        "employees": filtered_employees
    }

    return report


def print_department_report(report):
    """Pretty-print department report dictionary."""
    print_separator()
    print(report["report_title"])
    print_separator()
    print("Company Name           :", report["company_name"])
    print("Department             :", report["department"])
    print("Employee Count         :", report["employee_count"])
    print("Total Salary           :", report["total_salary"])
    print("Average Salary         :", report["average_salary"])
    print("Average Performance    :", report["average_performance_score"])
    print("\nEmployees")
    display_employees(report["employees"])


def main():
    while True:
        print("\nEMPLOYEE PERFORMANCE ANALYSIS SYSTEM")
        print_separator()
        print("1. Display all employees")
        print("2. Display employee names")
        print("3. Display high performers")
        print("4. Search by department")
        print("5. Calculate revised salaries")
        print("6. Calculate total salary expense")
        print("7. Find highest-paid employee")
        print("8. Sort employees")
        print("9. Generate department report")
        print("10. Exit")
        print_separator()

        choice = input("Enter your choice (1-10): ").strip()

        if choice == "1":
            print("\nAll Employees")
            display_employees(employees)

        elif choice == "2":
            print("\nEmployee Names")
            names = get_employee_names(employees)
            print(names)

        elif choice == "3":
            print("\nHigh Performers (Performance Score >= 80)")
            high_performers = get_high_performers(employees)
            display_employees(high_performers)

        elif choice == "4":
            dept = input("Enter department name: ").strip()
            matches = filter_by_department(employees, dept)

            if matches:
                print(f"\nEmployees in {dept.title()} Department")
                display_employees(matches)
            else:
                print(f"No employees found in {dept.title()} department.")

        elif choice == "5":
            revised = get_revised_salaries(employees)
            print("\nRevised Salaries Based on Performance")
            display_employees(revised)
            show_bonus_demo(employees)

        elif choice == "6":
            current_total, revised_total, additional = get_total_additional_expense(employees)
            print("\nSalary Expense Summary")
            print_separator()
            print("Current Total Salary      :", current_total)
            print("Revised Total Salary      :", round(revised_total, 2))
            print("Additional Salary Expense  :", round(additional, 2))
            print_separator()

        elif choice == "7":
            highest_paid = get_highest_paid_employee(employees)
            if highest_paid:
                print("\nHighest-Paid Employee")
                display_employees([highest_paid])
            else:
                print("Employee list is empty.")

        elif choice == "8":
            by_name, by_salary, by_perf, by_exp = sort_employees(employees)

            print("\nSorted by Name")
            display_employees(by_name)

            print("\nSorted by Salary")
            display_employees(by_salary)

            print("\nSorted by Performance Score")
            display_employees(by_perf)

            print("\nSorted by Experience")
            display_employees(by_exp)

        elif choice == "9":
            dept = input("Enter department name: ").strip()
            report = generate_department_report(employees, dept)
            print_department_report(report)

        elif choice == "10":
            print("Exiting program. Goodbye!")
            break

        else:
            print("Invalid choice. Please enter a number between 1 and 10.")


if __name__ == "__main__":
    main()
```
